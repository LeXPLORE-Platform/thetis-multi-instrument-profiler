```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv("data.csv", sep=" ")
df.head()
```

```python
def fit(landa, k, Sk):
    return k*(landa/532)**(-Sk)
curve_fit(fit, np.array(df["landa"]), np.array(df["C"]))[0][1]
```
